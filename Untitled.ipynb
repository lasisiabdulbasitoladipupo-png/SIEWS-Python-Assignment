```python
# Syntax: lambda parameters : expression
# A regular function that doubles a number
def double_regular(x):
 return x * 2
# The same function as a lambda
double_lambda = lambda x: x * 2
# Both work identically
print(double_regular(5)) # 10
print(double_lambda(5)) # 10
# Lambda with two parameters
add = lambda x, y: x + y
print(add(3, 7)) # 10
# Lambda with a conditional expression
classify = lambda score: 'Pass' if score >= 50 else 'Fail'
print(classify(72)) # Pass
print(classify(41)) # Fail
```

```python
# map(function, iterable) -- applies function to every element
# Example 1: Convert Celsius to Fahrenheit
celsius = [0, 20, 37, 100]
def to_fahrenheit(c):
 return (c * 9/5) + 32
fahrenheit = list(map(to_fahrenheit, celsius))
print(fahrenheit) # [32.0, 68.0, 98.6, 212.0]
# Example 2: Same thing with a lambda (no need to define a function)
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit) # [32.0, 68.0, 98.6, 212.0]
# Example 3: Apply to strings
names = [' amara ', 'KOFI ', ' fatima']
cleaned = list(map(str.strip, names)) # Remove whitespace
lowered = list(map(str.lower, cleaned)) # Lowercase
print(lowered) # ['amara', 'kofi', 'fatima']
# Example 4: map() with two iterables
prices = [100, 200, 150]
discounts = [0.1, 0.2, 0.15]
final_prices = list(map(lambda p, d: p * (1 - d), prices, discounts))
print(final_prices) # [90.0, 160.0, 127.5]
```

```python
# Scenario: You have raw exam scores.
# Step 1: Remove invalid scores (below 0 or above 100).
# Step 2: Scale valid scores from 0-100 to 0-1.
raw = [85, -5, 92, 101, 67, 78, 0, 45, 110]
# Filter: keep only scores in the valid range [0, 100]
valid = list(filter(lambda s: 0 <= s <= 100, raw))
print('Valid scores:', valid) # [85, 92, 67, 78, 0, 45]
# Map: scale each valid score to [0, 1]
scaled = list(map(lambda s: round(s / 100, 2), valid))
print('Scaled scores:', scaled) # [0.85, 0.92, 0.67, 0.78, 0.0, 0.45]
# All in one pipeline using list comprehension (equivalent, often preferred)
pipeline = [round(s / 100, 2) for s in raw if 0 <= s <= 100]
print('Pipeline:', pipeline) # [0.85, 0.92, 0.67, 0.78, 0.0, 0.45]
```

```python
import numpy as np
scores = np.array([85, -5, 92, 101, 67, 78, 0, 45, 110])
# NumPy equivalent of filter (boolean indexing)
valid = scores[(scores >= 0) & (scores <= 100)]
print('Valid:', valid) # [85 92 67 78 0 45]
# NumPy equivalent of map (vectorised operation)
scaled = valid / 100.0
print('Scaled:', scaled) # [0.85 0.92 0.67 0.78 0. 0.45]
# NumPy equivalent of map with a complex function (np.vectorize)
# Use this only when no vectorised NumPy operation exists
assign_grade = np.vectorize(lambda s: 'A' if s >= 90 else 'B' if s >= 80 else
'C')
grades = assign_grade(np.array([95, 82, 74]))
print(grades) # ['A' 'B' 'C']
```

```python
import numpy as np
# Raw data: student records with some issues
raw_records = [
 {'id': 'STU001', 'age': 22, 'score1': 78, 'score2': 85, 'label':
'pass'},
 {'id': 'STU002', 'age': 19, 'score1': -10, 'score2': 92, 'label':
'pass'},
 {'id': 'STU003', 'age': 25, 'score1': 55, 'score2': 48, 'label':
'fail'},
 {'id': 'STU004', 'age': 21, 'score1': 88, 'score2': 110, 'label':
'pass'},
 {'id': 'STU005', 'age': 23, 'score1': 60, 'score2': 70, 'label':
'pass'},
 {'id': 'STU006', 'age': 20, 'score1': 40, 'score2': 35, 'label':
'fail'},
]
# --- Step 1: Filter out records with out-of-range scores ---
# A valid score is between 0 and 100 inclusive.
is_valid = lambda r: 0 <= r['score1'] <= 100 and 0 <= r['score2'] <= 100
clean_records = list(filter(is_valid, raw_records))
print(f'Records before filter: {len(raw_records)}')
print(f'Records after filter : {len(clean_records)}')
# --- Step 2: Extract features using a list comprehension ---
# Features: [age, score1, score2, average_score]
features = [
 [r['age'],
 r['score1'],
 r['score2'],
 (r['score1'] + r['score2']) / 2]
 for r in clean_records
]
# --- Step 3: Extract labels and encode them with map + lambda ---
# pass -> 1, fail -> 0
```

```python
encode_label = lambda r: 1 if r['label'] == 'pass' else 0
labels = list(map(encode_label, clean_records))
print('Labels:', labels)
# --- Step 4: Convert to NumPy arrays ---
X = np.array(features, dtype=np.float32)
y = np.array(labels, dtype=np.int32)
print('Feature matrix shape:', X.shape) # (4, 4)
print('Label array shape: ', y.shape) # (4,)
# --- Step 5: Normalise each feature column to [0, 1] ---
col_min = X.min(axis=0) # Minimum per column
col_max = X.max(axis=0) # Maximum per column
X_scaled = (X - col_min) / (col_max - col_min + 1e-8) # 1e-8 avoids /0
print('Scaled features (first row):', X_scaled[0].round(2))
# --- Step 6: Print a summary using list comprehension ---
col_names = ['age', 'score1', 'score2', 'avg_score']
summary = [(name, round(X[:, i].mean(), 2), round(X[:, i].std(), 2))
 for i, name in enumerate(col_names)]
print('\nFeature Summary:')
print(f'{"Feature":<14} {"Mean":>8} {"Std":>8}')
for name, mean, std in summary:
 print(f'{name:<14} {mean:>8} {std:>8}')
```
